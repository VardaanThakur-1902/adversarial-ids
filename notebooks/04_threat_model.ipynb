# Phase 4 — Adversarial Threat Model

## Objective

Define the adversary capabilities, attack objective,
perturbation constraints, and evaluation methodology
for adversarial attacks against the IDS.

## Threat Model

The initial experiment uses a white-box attacker.

The attacker is assumed to know:
- IDS model architecture
- Model parameters
- Input features
- Loss function
- Model gradients

## Attack Objective

The primary objective is evasion:

ATTACK → BENIGN

The attacker attempts to modify malicious network-flow
features so that the IDS misclassifies the traffic as benign.

## Adversarial Attacks

1. FGSM
2. PGD

Both attacks operate in the standardized feature space.

## Perturbation Constraint

An L-infinity constraint is used initially.

The perturbation is bounded by epsilon:

|x_adv - x| <= epsilon

## Perturbation Strength

Initial epsilon values:

0.001
0.005
0.01
0.05
0.1

These values are applied to standardized features.

## Evaluation

For each attack configuration we measure:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC
- PR-AUC
- FPR
- FNR
- Attack Success Rate

Clean validation performance is used as the baseline.

## Research Limitation

Feature-space adversarial perturbations may not always
correspond to physically realizable network traffic.
Therefore, adversarial success in this experiment should
be interpreted as vulnerability in the learned feature
representation rather than automatic proof of a
real-world executable network attack.
