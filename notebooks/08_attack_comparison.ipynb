```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

RESULTS_DIR = Path("../experiments/results")

fgsm_df = pd.read_csv(
    RESULTS_DIR / "fgsm_results.csv"
)

pgd_df = pd.read_csv(
    RESULTS_DIR / "pgd_results.csv"
)

print("FGSM:")
display(fgsm_df)

print("\nPGD:")
display(pgd_df)
```

```python
clean_metrics = {
    "Accuracy": 0.983680,
    "Precision": 0.960077,
    "Recall": 0.942563,
    "F1": 0.951239,
    "ROC-AUC": 0.998706,
    "PR-AUC": 0.994001,
    "FPR": 0.007965,
    "FNR": 0.057437
}
```

```python
comparison = pd.DataFrame({
    "Epsilon": fgsm_df["Epsilon"],

    "FGSM Recall": fgsm_df["Recall"],
    "PGD Recall": pgd_df["Recall"],

    "FGSM F1": fgsm_df["F1"],
    "PGD F1": pgd_df["F1"],

    "FGSM FNR": fgsm_df["FNR"],
    "PGD FNR": pgd_df["FNR"],

    "FGSM ASR": fgsm_df["Attack Success Rate"],
    "PGD ASR": pgd_df["Attack Success Rate"]
})

comparison
```

```python
comparison["PGD - FGSM Recall"] = (
    comparison["PGD Recall"] -
    comparison["FGSM Recall"]
)
```

```python
comparison["PGD - FGSM F1"] = (
    comparison["PGD F1"] -
    comparison["FGSM F1"]
)
```

```python
comparison["PGD - FGSM FNR"] = (
    comparison["PGD FNR"] -
    comparison["FGSM FNR"]
)
```

```python
comparison["PGD - FGSM ASR"] = (
    comparison["PGD ASR"] -
    comparison["FGSM ASR"]
)
```

```python
comparison[
    [
        "Epsilon",
        "PGD - FGSM Recall",
        "PGD - FGSM F1",
        "PGD - FGSM FNR",
        "PGD - FGSM ASR"
    ]
]
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    fgsm_df["Epsilon"],
    fgsm_df["Recall"],
    marker="o",
    label="FGSM"
)

plt.plot(
    pgd_df["Epsilon"],
    pgd_df["Recall"],
    marker="s",
    label="PGD"
)

plt.axhline(
    clean_metrics["Recall"],
    linestyle="--",
    label="Clean"
)

plt.xlabel("Epsilon")
plt.ylabel("Recall")
plt.title(
    "Neural Network Recall: Clean vs FGSM vs PGD"
)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    fgsm_df["Epsilon"],
    fgsm_df["F1"],
    marker="o",
    label="FGSM"
)

plt.plot(
    pgd_df["Epsilon"],
    pgd_df["F1"],
    marker="s",
    label="PGD"
)

plt.axhline(
    clean_metrics["F1"],
    linestyle="--",
    label="Clean"
)

plt.xlabel("Epsilon")
plt.ylabel("F1-score")
plt.title(
    "Neural Network F1 Under Adversarial Attacks"
)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    fgsm_df["Epsilon"],
    fgsm_df["FNR"],
    marker="o",
    label="FGSM"
)

plt.plot(
    pgd_df["Epsilon"],
    pgd_df["FNR"],
    marker="s",
    label="PGD"
)

plt.axhline(
    clean_metrics["FNR"],
    linestyle="--",
    label="Clean"
)

plt.xlabel("Epsilon")
plt.ylabel("False Negative Rate")
plt.title(
    "False Negative Rate Under FGSM and PGD"
)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    fgsm_df["Epsilon"],
    fgsm_df["Attack Success Rate"],
    marker="o",
    label="FGSM"
)

plt.plot(
    pgd_df["Epsilon"],
    pgd_df["Attack Success Rate"],
    marker="s",
    label="PGD"
)

plt.xlabel("Epsilon")
plt.ylabel("Attack Success Rate")
plt.title(
    "FGSM vs PGD Attack Success Rate"
)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fgsm_degradation = pd.DataFrame({
    "Epsilon": fgsm_df["Epsilon"],

    "Recall Degradation":
        clean_metrics["Recall"] -
        fgsm_df["Recall"],

    "F1 Degradation":
        clean_metrics["F1"] -
        fgsm_df["F1"],

    "Accuracy Degradation":
        clean_metrics["Accuracy"] -
        fgsm_df["Accuracy"],

    "FNR Increase":
        fgsm_df["FNR"] -
        clean_metrics["FNR"],

    "Attack Success Rate":
        fgsm_df["Attack Success Rate"]
})

fgsm_degradation
```

```python
pgd_degradation = pd.DataFrame({
    "Epsilon": pgd_df["Epsilon"],

    "Recall Degradation":
        clean_metrics["Recall"] -
        pgd_df["Recall"],

    "F1 Degradation":
        clean_metrics["F1"] -
        pgd_df["F1"],

    "Accuracy Degradation":
        clean_metrics["Accuracy"] -
        pgd_df["Accuracy"],

    "FNR Increase":
        pgd_df["FNR"] -
        clean_metrics["FNR"],

    "Attack Success Rate":
        pgd_df["Attack Success Rate"]
})

pgd_degradation
```

```python
attack_comparison = pd.DataFrame({
    "Epsilon": fgsm_df["Epsilon"],

    "FGSM Accuracy": fgsm_df["Accuracy"],
    "PGD Accuracy": pgd_df["Accuracy"],

    "FGSM Recall": fgsm_df["Recall"],
    "PGD Recall": pgd_df["Recall"],

    "FGSM F1": fgsm_df["F1"],
    "PGD F1": pgd_df["F1"],

    "FGSM FNR": fgsm_df["FNR"],
    "PGD FNR": pgd_df["FNR"],

    "FGSM ASR": fgsm_df["Attack Success Rate"],
    "PGD ASR": pgd_df["Attack Success Rate"]
})

attack_comparison
```

```python
attack_comparison.to_csv(
    RESULTS_DIR /
    "fgsm_vs_pgd_comparison.csv",
    index=False
)

print("Attack comparison saved.")
```
